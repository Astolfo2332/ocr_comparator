```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
```

```python
test_data_path = "../data/test_dataset.csv"
test_df = pd.read_csv(test_data_path)
```

```python
empresas = test_df["PROVEEDOR"].value_counts()
empresas_encode = LabelEncoder()
test_df["PROVEEDOR_ENCODED"] = empresas_encode.fit_transform(test_df["PROVEEDOR"])
empresas_encoded = test_df["PROVEEDOR_ENCODED"].value_counts()
```

```python
plt.bar(empresas_encoded.index, empresas_encoded.values)
plt.title("Distribución de Proveedores en el Dataset de Prueba")
plt.xlabel("Proveedor")
plt.ylabel("Cantidad de Documentos")
plt.xticks(range(len(empresas_encoded.index)), empresas_encoded.index)
plt.tight_layout()
plt.grid(axis='y')
plt.show()
```

```python
#Rango de precios
plt.hist(test_df["TOTAL"] / 4000, bins=50, edgecolor='black')
plt.title("Distribución de Totales en el Dataset de Prueba USD")
plt.xlabel("Total")
plt.ylabel("Cantidad de Documentos")
plt.tight_layout()
plt.grid(axis='y')
plt.show()
```

```python
#Rango de iva
plt.hist(test_df["IVA"] / 4000, bins=50, edgecolor='black')
plt.title("Distribución de IVA en el Dataset de Prueba USD")
plt.xlabel("IVA")
plt.ylabel("Cantidad de Documentos")
plt.tight_layout()
plt.grid(axis='y')
plt.show()
```
