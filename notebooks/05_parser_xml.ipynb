```python
import xml.etree.ElementTree as ET
from bs4 import BeautifulSoup
import os
import pandas as pd
import datetime

main = os.getcwd().split("notebooks")[0]
test = os.path.join(main, "data", "extracted")
xml_file = [file for file in os.listdir(test) if file.endswith(".xml")][2]

print(f"Processing file: {os.path.join(test, xml_file)}")
root = ET.parse(os.path.join(test, xml_file)).getroot()
namespaces = {
    'xsi': "http://www.w3.org/2001/XMLSchema-instance",
    'xsd': "http://www.w3.org/2001/XMLSchema",
    'cac': "urn:oasis:names:specification:ubl:schema:xsd:CommonAggregateComponents-2",
    'cbc': "urn:oasis:names:specification:ubl:schema:xsd:CommonBasicComponents-2",
    'sts': "dian:gov:co:facturaelectronica:Structures-2-1",
    'xades': "http://uri.etsi.org/01903/v1.3.2#",
    'xades141': "http://uri.etsi.org/01903/v1.4.1#",
    'ext': "urn:oasis:names:specification:ubl:schema:xsd:CommonExtensionComponents-2",
    'ds': "http://www.w3.org/2000/09/xmldsig#",
    'inv': "urn:oasis:names:specification:ubl:schema:xsd:Invoice-2"
}

# Find and extract the PayableAmount value
payable_amount = root.find('.//cbc:PayableAmount', namespaces).text
payable_amount = float(payable_amount)

iva = root.find('.//cbc:TaxAmount', namespaces).text
iva = float(iva)

antes_iva = root.find('.//cbc:TaxExclusiveAmount', namespaces).text
antes_iva = float(antes_iva)

fecha = root.find('.//xades:SigningTime', namespaces).text.split("T")[0]
fecha = datetime.datetime.fromisoformat(fecha)
fecha = fecha.strftime("%#d/%m/%Y")

proveedor = root.find('.//cbc:Name', namespaces).text

nit = root.find('.//cbc:CompanyID', namespaces).text

factura = root.find('.//cbc:ID', namespaces).text


bills_df = pd.DataFrame(columns=["FECHA", "PROVEEDOR", "NIT", "FACTURA", "VALOR ANTES DE IVA", "IVA", "TOTAL"])

bills_df.loc[len(bills_df)] = [fecha, proveedor, nit, factura, antes_iva, iva, payable_amount]

bills_df


```
