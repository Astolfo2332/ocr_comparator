```python
for file in df_only_data["FILE_NAME"]:
    src = os.path.join("../data/extracted", file)
    dst = os.path.join("../data/test_dataset_pdfs", file)
    if os.path.exists(src):
        os.rename(src, dst)
```

```python
def pdf_to_img(pdf_path):
    folder = os.path.dirname(pdf_path)
    filename = os.path.splitext(os.path.basename(pdf_path))[0]

    images = convert_from_path(pdf_path, dpi=100)

    for i, img in enumerate(images):
        img_filename = f"{filename}_page_{i+1}.jpg"
        if os.path.exists(os.path.join(folder, img_filename)):
            continue
        img_path = os.path.join(folder, img_filename)

        # Save the image in JPG format
        img.save(img_path, "JPEG")
        print(f"Saved: {img_path}")
```

```python
main = os.getcwd().split("notebooks")[0]

for file in os.listdir("../data/test_dataset_pdfs"):
    if file.endswith(".pdf"):
        pdf_path = os.path.join(main, "data", "test_dataset_pdfs", file)
        pdf_to_img(pdf_path)
```
